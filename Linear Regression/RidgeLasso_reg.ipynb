### Algerian forest fire dataset

```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df=pd.read_csv("Algerian_forest_fires_dataset_UPDATE.csv",header =1 )
df.head(5)
```

```python
df.info()
```

Data cleaning

```python
df[df.isnull().any(axis=1)]
```

The dataset is actually divided into 2 nd set from the index 122 
1. bejaia region dataset
2. sidi bel abbes region dataset

add new coloumn with region

```python
df.loc[:122,'Region']=0
df.loc[122:,"Region"]=1
```

```python
df.info()
```

```python
df[['Region']]=df[['Region']].astype(int)
```

```python
##removing the null values
df=df.dropna().reset_index(drop=True)
```

```python
df.isnull().sum()
```

```python
df.iloc[[122]]
```

```python
df=df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.columns
#we can see some column name have extra space 
```

```python
#fix spaces in columns 
df.columns=df.columns.str.strip()
df.columns
```

```python
#change req columns to suitable data types

df[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']]=df[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']].astype(int)
```

```python
#changing the other columns to float 
objects=[features for features in df.columns if df[features].dtypes=='O']
```

```python
for i in objects:
    if i != 'Classes':
        df[i]=df[i].astype(float)
```

```python
df.info()
```

```python
df.describe()
```

```python
##kets save the cleaned data set
df.to_csv('Algerian_forest_fires_clean.csv',index=False)
```
