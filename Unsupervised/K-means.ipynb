### **K means**

```python
import pandas as pd 
import numpy as np
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
X,y = make_blobs(n_samples=1000,centers=3,n_features=2)
```

```python
X
```

```python
y# i have 3 centres thats why i got labels as 0,1,2
```

```python
plt.scatter(X[:,0],X[:,1],c=y)
```

```python
#standardization--features scaling techniques
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y, test_size=0.33,random_state=42)
```

```python
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
from sklearn.cluster import KMeans
```

```python
#Elbow method to select K Value
wcss=[]
for k in range(1,11):
    kmeans=KMeans(n_clusters=k,init='k-means++')
    kmeans.fit(X_train_scaled)
    wcss.append(kmeans.inertia_)
```

```python
wcss #value for 1 to 10
```

```python
#plot elbow curve
plt.plot(range(1,11),wcss)
plt.xticks(range(1,11))
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.show()
```

Our goal is to find the point from where the value abruptly decreases on the elbow 

```python
kmeans=KMeans(n_clusters=3,init='k-means++')
```

```python
kmeans.fit_predict(X_train_scaled)
```

```python
y_pred=kmeans.predict(X_test_scaled)
```

```python
plt.scatter(X_test[:,0],X_test[:,1],c=y_pred)
```

```python
#validating the k values using two different technique
#kneelocator
#silhoutee scoring 
```

```python
#kneelocator

from kneed import KneeLocator
```

```python
kl=KneeLocator(range(1,11),wcss,curve='convex',direction='decreasing')
```

```python
kl.elbow
```

```python
#silhoutee scoring 
from sklearn.metrics import silhouette_score
```

```python
silhouette_coefficients=[]
for k in range (2,11):
    kmeans=KMeans(n_clusters=k,init='k-means++')
    kmeans.fit(X_train_scaled)
    score=silhouette_score(X_train_scaled,kmeans.labels_)
    silhouette_coefficients.append(score)
```

```python
silhouette_coefficients
```

```python
#plotting silhouette score
plt.plot(range(2,11),silhouette_coefficients)
plt.xticks(range(2,11))
plt.xlabel('Number of clusters')
plt.ylabel('silhouette_coefficients')
plt.show()
```
